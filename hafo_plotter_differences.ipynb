```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
input_file = 'for011'

r_list = []
rho_charge_list = [] # this is referred to as rho-coul in the data file.
rho_proton_list = []
rho_neutron_list = []
rho_mass_list = []

q_list = []
ff_charge_list = [] # same note as rho_charge_list
ff_proton_list = []
ff_neutron_list = []   
ff_total_list = []
ff_weak_list = []

div_j_proton_list = []
div_j_neutron_list = []


with open(input_file, 'r') as f:
    lines = f.readlines()

for i in range ( 45 , 115 + 1 ):

    index,r,rho_charge,rho_proton,rho_neutron,rho_mass, fill1, fill2 , fill3, fill4 = lines[i].split()

    r_list.append(float(r))
    rho_charge_list.append(float(rho_charge))
    rho_proton_list.append(float(rho_proton))
    rho_neutron_list.append(float(rho_neutron))
    rho_mass_list.append(float(rho_mass))
    
for i in range ( 119 , 189 + 1 ) :

    index,q,ff_charge,ff_proton,ff_neutron,ff_total,ff_weak = lines[i].split()

    q_list.append(float(q))
    ff_charge_list.append(float(ff_charge))
    ff_proton_list.append(float(ff_proton))
    ff_neutron_list.append(float(ff_neutron))
    ff_total_list.append(float(ff_total))
    ff_weak_list.append(float(ff_weak))

for i in range ( 265 , 335 + 1 ) :

    div_j_neutron = lines[i].split()[-1]
    div_j_proton = lines[i].split()[-2]

    div_j_proton_list.append(float(div_j_proton))
    div_j_neutron_list.append(float(div_j_neutron))
```

```python
plt.plot(r_list, rho_charge_list, label=r'$\rho_{charge}$')
plt.plot(r_list, rho_proton_list, label=r'$\rho_{proton}$')
plt.plot(r_list, rho_neutron_list, label=r'$\rho_{neutron}$')
plt.plot(r_list, rho_mass_list, label=r'$\rho_{mass}$')

plt.xlabel('r [ fm ]')
plt.ylabel(r'$\rho$ [ Unit Measure $\times$ fm$^{-3}$ ]')

plt.grid( ls = '--' , lw = 0.5 )
plt.tick_params(direction='in')
plt.legend();
```

```python
plt.plot(q_list, ff_charge_list, label=r'$F_{charge}$',marker='s')
plt.plot(q_list, ff_proton_list, label=r'$F_{proton}$',marker='x')
plt.plot(q_list, ff_neutron_list, label=r'$F_{neutron}$')
plt.plot(q_list, ff_total_list, label=r'$F_{total}$')
plt.plot(q_list, ff_weak_list, label=r'$F_{weak}$')

plt.xlim(0, 2)

plt.xlabel('q')
plt.ylabel(r'F, Form Factor')

plt.grid(ls='--', alpha=0.5)
plt.tick_params(direction='in')
plt.legend();
```

```python
plt.plot(r_list, ff_charge_list, label=r'$F_{charge}$',marker='s')
plt.plot(r_list, ff_proton_list, label=r'$F_{proton}$',marker='x')
plt.plot(r_list, ff_neutron_list, label=r'$F_{neutron}$')
plt.plot(r_list, ff_total_list, label=r'$F_{total}$')
plt.plot(r_list, ff_weak_list, label=r'$F_{weak}$')

plt.xlim(0, 7)

plt.xlabel('r [ fm ]')
plt.ylabel(r'F, Form Factor')

plt.grid(ls='--', alpha=0.5)
plt.tick_params(direction='in')
plt.legend();
```

```python
plt.plot(r_list, div_j_neutron_list, label=r'$(\nabla \cdot j)_{neutron}$')
plt.plot(r_list, div_j_proton_list, label=r'$(\nabla \cdot j)_{proton}$')


plt.ylabel(r'$\nabla \cdot j$ [ $s^{-1}m^{-3}$ ]')
plt.xlabel('r [ fm ]')

plt.xlim(0, 11)

plt.grid(ls='--', alpha=0.5)
plt.tick_params(direction='in')
plt.legend();
```
