```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
input_file1 = 'fort.20'
input_file2 = 'for011'

r_list = []
rho_charge_list = [] # this is referred to as rho-coul in the data file.
rho_proton_list = []
rho_neutron_list = []
rho_mass_list = []

tau_neutron_list = []
tau_proton_list = []

sjs_proton_list = []
sjs_neutron_list = []


with open(input_file1, 'r') as f:
    lines1 = f.readlines()
with open(input_file2, 'r') as f:
    lines2 = f.readlines()

for i in range ( 4 , 74 + 1 ):

    index,r,rho_charge,rho_proton,rho_neutron,rho_mass = lines1[i].split()

    r_list.append(float(r))
    rho_charge_list.append(float(rho_charge))
    rho_proton_list.append(float(rho_proton))
    rho_neutron_list.append(float(rho_neutron))
    rho_mass_list.append(float(rho_mass))

for i in range ( 49 , 119 + 1 ) :

    sjs_neutron = lines2[i].split()[-1]
    sjs_proton = lines2[i].split()[-2]
    tau_neutron = lines2[i].split()[-3]
    tau_proton = lines2[i].split()[-4]

    sjs_proton_list.append(float(sjs_proton))
    sjs_neutron_list.append(float(sjs_neutron))
    tau_proton_list.append(float(tau_proton))
    tau_neutron_list.append(float(tau_neutron))

r_array = np.array(r_list)

rho_charge_array = np.array(rho_charge_list)
rho_proton_array = np.array(rho_proton_list)
rho_neutron_array = np.array(rho_neutron_list)
rho_mass_array = np.array(rho_mass_list)

tau_neutron_array = np.array(tau_neutron_list)
tau_proton_array = np.array(tau_proton_list)

sjs_proton_array = np.array(sjs_proton_list)
sjs_neutron_array = np.array(sjs_neutron_list)
```

```python
plt.plot(r_array, rho_charge_array, label=r'$\rho_{charge}$', color='blue')
plt.plot(r_array, rho_proton_array, label=r'$\rho_{proton}$', color='red')
plt.plot(r_array, rho_neutron_array, label=r'$\rho_{neutron}$', color='green')
plt.plot(r_array, rho_mass_array, label=r'$\rho_{mass}$', color='black')

plt.xlabel(r'$r$ [ fm ]')
plt.ylabel(r'$\rho$ [ Unit Measure fm$^{-3}$ ]')

plt.xlim(0, 7)

plt.tick_params(direction='in')
plt.grid(ls='--',alpha=0.5)
plt.legend();
```

```python
plt.plot(r_array, tau_neutron_array, label=r'$\tau_{neutron}$', color='blue')
plt.plot(r_array, tau_proton_array, label=r'$\tau_{proton}$', color='red')

plt.xlabel(r'$r$ [ fm ]')
plt.ylabel(r'$\tau$ [ MeV fm$^{-3}$ ]')

plt.xlim(0, 12)

plt.tick_params(direction='in')
plt.grid(ls='--',alpha=0.5)
plt.legend();
```

```python
fig, axs = plt.subplots(1, 2, figsize=(9, 4))  # 1 row, 2 columns

axs[0].plot(r_array, sjs_neutron_array, label=r'$\nabla\cdot J_{ls, neutron}$', color='blue')
axs[0].legend()
axs[0].set_xlabel(r'$r$ [ fm ]')
axs[0].set_ylabel(r'$\nabla\cdot J_{ls}$ [ MeV fm$^{-3}$ ]')
axs[0].set_xlim(0, 12)
axs[0].tick_params(direction='in')
axs[0].grid(ls='--', alpha=0.5)

axs[1].plot(r_array, sjs_proton_array, label=r'$\nabla\cdot J_{ls, proton}$', color='green')
axs[1].legend()
axs[1].set_xlabel(r'$r$ [ fm ]')
axs[1].set_ylabel(r'$\nabla\cdot J_{ls}$ [ MeV fm$^{-3}$ ]')
axs[1].set_xlim(0, 12)
axs[1].tick_params(direction='in')
axs[1].grid(ls='--', alpha=0.5)

plt.tight_layout();
```

```python
file = 'j3_quants.txt'

r,rp,rm,tp,tm,ssp,ssm,sp,sm = np.loadtxt ( file , dtype = float , unpack=True )

plt.plot(r,rp,label=r'$\rho_p$')
plt.plot(r,rm,label=r'$\rho_n$')
plt.plot(r,rm+rp,label=r'$\rho_0$')

plt.xlabel(r'$r$ [ fm ]')
plt.ylabel(r'$\rho$ [ Unit Measure fm$^{-3}$ ]')

plt.xlim(0, 7)

plt.tick_params(direction='in')
plt.grid(ls='--',alpha=0.5)
plt.legend();
```

```python
plt.plot(r,tp,label=r'$\tau_p$')
plt.plot(r,tm,label=r'$\tau_n$')
plt.plot(r,tm+tp,label=r'$\tau_0$')

plt.xlabel(r'$r$ [ fm ]')
plt.ylabel(r'$\tau$ [ MeV fm$^{-3}$ ]')

plt.xlim(0, 7)

plt.tick_params(direction='in')
plt.grid(ls='--',alpha=0.5)
plt.legend();
```

```python
plt.plot(r,sp,label=r'$J_p$')
plt.plot(r,sm,label=r'$J_n$')
plt.plot(r,sm+sp,label=r'$J_0$')

plt.xlabel(r'$r$ [ fm ]')
plt.ylabel(r'$J$')

plt.xlim(0, 10)

plt.tick_params(direction='in')
plt.grid(ls='--',alpha=0.5)
plt.legend();
```

```python
plt.plot(r,ssp,label=r'$\nabla\cdot J_p$')
plt.plot(r,ssm,label=r'$\nabla\cdot J_n$')
plt.plot(r,ssm+ssp,label=r'$\nabla\cdot J_0$')

plt.xlabel(r'$r$ [ fm ]')
plt.ylabel(r'$\nabla\cdot J$')

plt.xlim(0, 10)

plt.tick_params(direction='in')
plt.grid(ls='--',alpha=0.5)
plt.legend();
```
